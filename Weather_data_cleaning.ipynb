```python
EXTRA_ICAO = {'ADK': 'PADK', 'ADQ': 'PADQ', 'AKN': 'PANK', 'ANC': 'PANC', 'AZA': 'KIWA', 'BET': 'PABE', 'BQN': 'TJBQ', 'BRW': 'PABR', 'CDV': 'PACV', 'CLD': 'KCRQ', 'DLG': 'PADL', 'FAI': 'PAFA', 'FCA': 'KGPI', 'GST': 'PAGS', 'GUF': 'KJKA', 'GUM': 'PGUM', 'HHH': 'KHXD', 'HNL': 'PHNL', 'ITO':'PHTO', 'JNU': 'PAJN', 'KOA': 'PHKO', 'KTN': 'PAKT', 'LIH': 'PHLI', 'OGG': 'PHOG', 'OME': 'PAOM', 'OTZ': 'PAOT', 'PPG': 'NSTU', 'PSE': 'TJPS', 'PSG': 'PAPG', 'SCC': 'PASC', 'SCE': 'KUNV', 'SIT': 'PASI', 'SJU': 'TJSJ', 'SPN': 'PGSN', 'STT': 'TIST', 'STX': 'TISX', 'USA': 'KJQF', 'WRG': 'PAWG','YAK':'PAYA', 'YUM':'KNYL'} #not matched when I first ran this cell (non mainland airports)

stations = pd.read_csv('isd-history (1).csv', dtype = str) #make everything a string so that zeros are kept

stations['airport_code'] = stations['ICAO'].str[1:] #ICAO code - first letter

stations.loc[~stations['ICAO'].str.startswith('K', na=False)] #mainland airports

for iata, icao in EXTRA_ICAO.items():
  stations.loc[stations['ICAO'] == icao, 'airport_code'] = iata

stations = stations[stations['airport_code'].isin(airports)] #keep only airports that are in the list from Anna

stations = stations[stations['END'] >= '20250630'] #keep only stations that keep recording data after jun 30, 2025

stations = stations[stations['WBAN'] != '99999'] #keep only with a wban number

stations['station_id'] = stations['USAF'] + stations['WBAN'] #add the two ids together

stations = stations.drop_duplicates(subset = 'airport_code') #only one weather station should correspond to one airport

note(f'Airports with no identified weather station: {sorted(set(airports) - set(stations['airport_code']))}')
stations[['airport_code', 'station_id', 'STATION NAME', 'LAT', 'LON']]
```

```python
URL = 'https://noaa-global-hourly-pds.s3.amazonaws.com/2025/'
columns_to_keep = ['STATION', 'DATE', 'NAME', 'LATITUDE', 'LONGITUDE', 'REPORT_TYPE', 'WND', 'VIS', 'TMP', 'AA1']
months_to_keep = ['01', '02', '06', '07']
all_data = []

for _, row in stations.iterrows():
  url = URL + row['station_id'] + '.csv' #file address

  try:
    df = pd.read_csv(url, dtype = str, usecols=lambda c:c in columns_to_keep)
  except Exception:
    print(row['airport_code'], '- could not download, skipping')
    continue

  df = df[df['DATE'].str[5:7].isin(months_to_keep)]
  df['airport_code'] = row['airport_code'] #lable each row with the airport code
  all_data.append(df)

  print(row['airport_code'], '-', len(df), 'rows')

  raw = pd.concat(all_data, ignore_index = True) #add all airports to one table

  note(f'Rows before cleaning: {len(raw)}')
```

```python
raw.columns = raw.columns.str.lower()

raw = raw.rename(columns = {'station': 'station_id', 'name': 'station_name', 'date': 'date_utc'}) #rename a few columns to make them clearer when data is merged

note('Blank values per column (before cleaning):\n' + raw.isna().sum().to_string())

num_duplicates = raw.duplicated().sum()
note(f'Number of dropped duplicate rows: {num_duplicates}')

before = len(raw)
raw = raw[raw['report_type'].str.strip() == 'FM-15'] #only routine hourly reports are kept

note(f'Rows dropped because they were not routine reports: {before - len(raw)}')
```

```python
raw['wind_speed_ms'] = pd.to_numeric(raw['wnd'].str.split(',').str[3], errors='coerce')
raw['visibility_m'] = pd.to_numeric(raw['vis'].str.split(',').str[0], errors='coerce')
raw['temp_c'] = pd.to_numeric(raw['tmp'].str.split(',').str[0], errors='coerce')
raw['precip_hours'] = pd.to_numeric(raw['aa1'].str.split(',').str[0], errors='coerce')
raw['precip_mm'] = pd.to_numeric(raw['aa1'].str.split(',').str[1], errors='coerce')

raw[['wnd', 'wind_speed_ms', 'vis', 'visibility_m', 'tmp', 'temp_c', 'aa1', 'precip_mm']].head()
```

```python
missing_codes = {'wind_speed_ms': 9999, 'visibility_m': 999999, 'temp_c': 9999, 'precip_mm': 9999}

for column, code in missing_codes.items():
  count = (raw[column] == code).sum()
  note(f'{column}: {count} missing-value codes ({code}) replaced with a blank')
  raw[column] = raw[column].replace(code, np.nan)

raw['precip_mm'] = raw['precip_mm'].where(raw['precip_hours'] == 1) #only keeps the rain if its 1 hour long

raw['wind_speed_ms'] = raw['wind_speed_ms']/10
raw['temp_c'] = raw['temp_c']/10
raw['precip_mm'] = raw['precip_mm']/10

raw[['wind_speed_ms', 'visibility_m', 'temp_c', 'precip_mm']].head()
```

```python
tf = TimezoneFinder()

timezones = {}

for _, row in stations.iterrows():
  timezones[row['airport_code']] = tf.timezone_at(lat=float(row['LAT']), lng=float(row['LON'])) #Find time zones from location

print(timezones)

raw['date_utc'] = pd.to_datetime(raw['date_utc'], utc = True)

pieces = []

for airport, group in raw.groupby('airport_code'):
  group = group.copy()
  group['local_time'] = group['date_utc'].dt.tz_convert(timezones[airport]).dt.tz_localize(None)

  pieces.append(group)
raw = pd.concat(pieces)

raw['weather_date'] = raw['local_time'].dt.normalize()

raw['weather_hour'] = raw['local_time'].dt.hour

before = len(raw)

raw = raw[raw['local_time'].dt.month.isin([1, 6])] #only june and january should be kept

note(f'Rows dropped because they were not in January or June : {before - len(raw)}')

raw[['airport_code', 'date_utc', 'local_time', 'weather_date', 'weather_hour']].head()
```

```python
before = len(raw)
raw = raw.sort_values('date_utc')

weather = raw.drop_duplicates(subset = ['airport_code', 'weather_date', 'weather_hour'], keep='first')

note(f'Rows dropped so there is one row per airport for each hour: {before - len(weather)}')

weather = weather[['airport_code', 'station_id', 'station_id', 'station_name', 'latitude', 'longitude', 'weather_date', 'weather_hour', 'wind_speed_ms', 'visibility_m', 'temp_c', 'precip_mm']]
weather = weather.sort_values(['airport_code', 'weather_date', 'weather_hour'])

note(f'Rows after cleaning: {len(weather)}')

note('Blank values per column after cleaning:\n' + weather.isna().sum().to_string())

weather.head(10)
```

```python
weather.to_csv('weather_clean.csv', index = False, date_format = '%Y-%m-%d')

header = '''WEATHER DATA CLEANING NOTES
Source: NOAA Global Hourly (https://noaa-global-hourly-pds.s3.amazonaws.com/2025/) and NOAA (isd-history.csv)
Scope: January and June 2025
'''

with open('weather_notes.txt', 'w') as f:
  f.write(header + '\n\n'.join(notes))

from google.colab import files
files.download('weather_clean.csv')
files.download('weather_notes.txt')

from google.colab import drive
drive.mount('/content/drive')

import shutil
shutil.copy('weather_clean.csv', '/content/drive/MyDrive/weather_clean.csv')
shutil.copy('weather_notes.txt', '/content/drive/MyDrive/weather_notes.txt')

print('saved to google drive')
```
